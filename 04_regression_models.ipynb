### Linear Regression

```python
import numpy as np

class LinearRegressionScratch:
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.lr = learning_rate
        self.epochs = epochs
        self.w = None
        self.b = None
    
    def fit(self, X, y):
        n_samples = X.shape[0]
        # Step 1: initialise w and b (to 0 or small random values)
        self.w = 0 # initial value of w
        self.b = 0
        
        # Step 2: loop over the epochs
        for epoch in range(self.epochs):
            # a) Calculate predictions: y_pred = w*X + b
            y_pred = self.w*X + self.b
            
            # b) Calculate the gradients (formula above)
            dw = (-2/n_samples) * np.sum(X * (y - y_pred))
            db = (-2/n_samples) *np.sum(y - y_pred)
            
            # c) Update w and b
            self.w = self.w - self.lr * dw
            self.b = self.b - self.lr * db
    
    def predict(self, X):
        return self.w * X + self.b
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data (with some noise)
np.random.seed(42)
X = np.linspace(0, 10, 50)
y = 2*X + 1 + np.random.normal(0, 1, 50)  # true relationship: y = 2x + 1, with noise

# Train the model
print('Learning rate = 0.01')
model = LinearRegressionScratch(learning_rate=0.01, epochs=1000)
model.fit(X, y)

print(f"Learned w: {model.w:.4f}")   # Should be close to 2
print(f"Learned b: {model.b:.4f}")   # Should be close to 1

# Plot the predictions
y_pred = model.predict(X)

plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, color='red', label="Fitted Line")
plt.legend()
plt.show()
```

### Polynomial Regression

```python
import numpy as np
import matplotlib.pyplot as plt

# New synthetic data: quadratic relationship
np.random.seed(42)
X = np.linspace(-3, 3, 50)
y = 2*X**2 + 3*X + 5 + np.random.normal(0, 2, 50)  # true relationship: y = 2x² + 3x + 5

# TASK 1: feature engineering
# build X_poly with two columns: X and X²
X_poly = np.column_stack([X, X**2])

# Hint: you can use np.column_stack(): np.column_stack([X, X**2])
```

```python
class LinearRegressionScratch:
    def __init__(self, learning_rate = 0.01, epochs = 1000):
        self.lr = learning_rate
        self.epochs = epochs
        self.w = None
        self.b = None

    def fit(self,X, y):
        n_samples = X.shape[0]
        self.w = [0, 0]  # Initialize weights for two features (X and X²)
        self.b = 0

        for epoch in range(self.epochs):
            y_pred = X @ self.w + self.b

            dw = (-2/n_samples) * (X.T @ (y - y_pred))
            db = (-2/n_samples) * np.sum(y - y_pred)

            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        return X @ self.w + self.b
```

### Ridge Regression

```python
# Ridge Regression Scratch implementation
class RidgeRegressionScratch:
    def __init__(self, learning_rate=0.01, epochs=1000, lambda_=1.0):
        self.lr = learning_rate
        self.epochs = epochs
        self.lambda_ = lambda_
        self.w = None
        self.b = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)  # weights array for multiple features
        self.b = 0

        for epoch in range(self.epochs):
            y_pred = X @ self.w + self.b

            dw = (-2/n_samples) * (X.T @ (y - y_pred)) + 2 * self.lambda_ * self.w
            db = (-2/n_samples) * np.sum(y - y_pred)
    

            self.w -= self.lr * dw
            self.b -= self.lr * db

    def predict(self, X):
        return X @ self.w + self.b
```

### Lasso Regression

```python
# Lasso Regression Scratch implementation
class LassoRegressionScratch:
    def __init__(self, learning_rate=0.01, epochs=1000, lambda_=1.0):
        self.lr = learning_rate
        self.epochs = epochs
        self.lambda_ = lambda_
        self.w = None
        self.b = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)  # weights array for multiple features
        self.b = 0

        for epoch in range(self.epochs):
            y_pred = X @ self.w + self.b

            gradient = (-2/n_samples) * np.sum(X[:, i] * (y - y_pred))

            for i in range(n_features):

                if self.w[i] > 0:
                    dw = gradient + self.lambda_
                elif self.w[i] < 0:
                    dw = gradient - self.lambda_
                else:
                    if gradient > self.lambda_:
                        dw = gradient - self.lambda_
                    elif gradient < -self.lambda_:
                        dw = gradient + self.lambda_
                    else:
                        dw = 0
                self.w[i] -= self.lr * dw
         
            db = (-2/n_samples) * np.sum(y - y_pred)
    
            self.b -= self.lr * db

    def predict(self, X):
        return X @ self.w + self.b
```

### Decesion Tree Regression

```python
import numpy as np

def calculate_variance(y):
    if len(y) == 0:
        return 0
    return np.var(y)  # NumPy's built-in variance function

def weighted_variance(y_left, y_right):
    n_left, n_right = len(y_left), len(y_right)
    n_total = n_left + n_right
    
    weighted_var = (n_left/n_total) * calculate_variance(y_left) + (n_right/n_total) * calculate_variance(y_right)
    return weighted_var
```

```python
y_left = np.array([20, 22])
y_right = np.array([55, 58, 80, 85])

variance = weighted_variance(y_left, y_right)
variance
```

```python
def find_best_split(X, y):
    n_samples, n_features = X.shape
    best_variance = float('inf')
    best_feature = None
    best_threshold = None
    
    for feature_idx in range(n_features):
        thresholds = np.unique(X[:, feature_idx])  # all unique values of this feature
        
        for threshold in thresholds:
            left_mask = X[:, feature_idx] <= threshold
            right_mask = X[:, feature_idx] > threshold
            
            y_left, y_right = y[left_mask], y[right_mask]
            
            if len(y_left) == 0 or len(y_right) == 0:
                continue  # invalid split, skip it
            
            var = weighted_variance(y_left, y_right)
            
            if var < best_variance:
                best_variance = var
                best_feature = feature_idx
                best_threshold = threshold
    
    return best_feature, best_threshold, best_variance
```

```python
X = np.array([[1000], [1200], [1800], [2000], [2500], [2800]])  # sqft
y = np.array([20, 22, 55, 58, 80, 85])  # price in lakhs

feature, threshold, var = find_best_split(X, y)
print(f"Best feature index: {feature}, Best threshold: {threshold}, Variance: {var:.2f}")
```

```python
class Node:
    def __init__(self, feature=None, threshold=None, left=None, right=None, value=None):
        self.feature = feature      # feature used for the split (None for a leaf)
        self.threshold = threshold  # threshold used for the split (None for a leaf)
        self.left = left            # left child node (None for a leaf)
        self.right = right          # right child node (None for a leaf)
        self.value = value          # prediction value (leaf nodes only)
    
    def is_leaf(self):
        return self.value is not None
```

```python
def build_tree(X, y, depth=0, max_depth=5, min_samples_split=2):
    n_samples = X.shape[0]
    
    # STOPPING CONDITIONS: if any of these is true, make a leaf
    if (depth >= max_depth) or (n_samples < min_samples_split) or (len(np.unique(y)) == 1):
        leaf_value = np.mean(y)  # the leaf's prediction = the average of that group
        return Node(value=leaf_value)
    
    # Find the best split (built earlier)
    feature, threshold, variance = find_best_split(X, y)
    
    if feature is None:  # no valid split found
        return Node(value=np.mean(y))
    
    # Split the data
    left_mask = X[:, feature] <= threshold
    right_mask = X[:, feature] > threshold
    
    # RECURSION: call the same function again on the left and right subsets
    left_child = build_tree(X[left_mask], y[left_mask], depth+1, max_depth, min_samples_split)
    right_child = build_tree(X[right_mask], y[right_mask], depth+1, max_depth, min_samples_split)
    
    return Node(feature=feature, threshold=threshold, left=left_child, right=right_child)
```

```python
X = np.array([[1000], [1200], [1800], [2000], [2500], [2800]])
y = np.array([20, 22, 55, 58, 80, 85])

tree = build_tree(X, y, max_depth=3)
```

```python
def predict_single(node, x):
    if node.is_leaf():
        return node.value
    
    if x[node.feature] <= node.threshold:
        return predict_single(node.left, x)
    else:
        return predict_single(node.right, x)
```

### Random Forest

```python
import numpy as np

class RandomForestScratch:
    def __init__(self, n_trees=10, max_depth=5, max_features=None):
        self.n_trees = n_trees
        self.max_depth = max_depth
        self.max_features = max_features  # number of features to consider at each split
        self.trees = []  # all the trees are stored here
        self.tree_features = []  # features used by each tree (for tracking)
    
    def bootstrap_sample(self, X, y):
        n_samples = X.shape[0]
        indices = np.random.choice(n_samples, size=n_samples, replace=True)  # "with replacement"
        return X[indices], y[indices]
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        
        if self.max_features is None:
            self.max_features = n_features // 3  # typical default for regression
        
        self.trees = []
        self.tree_features = []
        
        for i in range(self.n_trees):
            X_sample, y_sample = self.bootstrap_sample(X, y)
            
            feature_indices = np.random.choice(n_features, size= self.max_features, replace= False)
            
            X_sample_subset = X_sample[:, feature_indices]  # X_sample[:, feature_indices]
            
            tree = build_tree(X_sample_subset, y_sample, max_depth=self.max_depth)
            
            self.trees.append(tree)
            self.tree_features.append(feature_indices)
    
    def predict(self, X):
        # Take a prediction from each tree, then average
        all_predictions = []
        
        for tree, feature_indices in zip(self.trees, self.tree_features):
            X_subset = X[:, feature_indices]
            tree_preds = np.array([predict_single(tree, x) for x in X_subset])
            all_predictions.append(tree_preds)
        
        all_predictions = np.array(all_predictions)  # shape: (n_trees, n_samples)
        return np.mean(all_predictions, axis=0)  # average across trees
```

### Gradient Boosting

```python
# Gradient Boosting from Scratch

import numpy as np

class GradientBoostingScratch:
    def __init__(self, n_estimators=50, learning_rate=0.1, max_depth=3):
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.max_depth = max_depth
        self.trees = []           # all the trees are stored here
        self.initial_prediction = None
    
    def fit(self, X, y):
        # Step 1: Initial prediction = average of y
        self.initial_prediction = np.mean(y)
        
        # Current predictions (initially the same for everyone: the average)
        current_predictions = np.full(y.shape, self.initial_prediction)
        
        self.trees = []
        
        for i in range(self.n_estimators):
            # Step 1: calculate the residuals
            residuals = y - current_predictions  # hint: y - current_predictions
            
            # Step 2: train a new tree with the residuals as the target
            tree = build_tree(X, residuals, max_depth=self.max_depth)
            
            # Step 3: get this tree's predictions on the training data
            tree_predictions = np.array([predict_single(tree, x) for x in X])  # hint: [predict_single(tree, x) for x in X], then np.array()
            
            # Step 4: update current_predictions
            current_predictions = current_predictions+ self.learning_rate * tree_predictions  # hint: current_predictions + learning_rate * tree_predictions
            
            self.trees.append(tree)
    
    def predict(self, X):
        # Prediction = initial_prediction + sum of (learning_rate * each tree's prediction)
        predictions = np.full(X.shape[0], self.initial_prediction)
        
        for tree in self.trees:
            tree_preds = np.array([predict_single(tree, x) for x in X])
            predictions += self.learning_rate * tree_preds
        
        return predictions
```

```python
X = np.array([[1000], [1200], [1800], [2000], [2500], [2800]])
y = np.array([20, 22, 55, 58, 80, 85])

gb_model = GradientBoostingScratch(n_estimators=50, learning_rate=0.1, max_depth=2)
gb_model.fit(X, y)

predictions = gb_model.predict(X)
print("Predictions:", predictions)
print("Actual:", y)
```
